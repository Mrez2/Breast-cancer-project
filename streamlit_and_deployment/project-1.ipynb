##### Step 5: Split Data 
##### - Split dataset into training and testing sets (70/30 split).

```python
X = df.drop('target', axis=1)
y = df['target']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

##### Step 6: Preprocessing 
##### - Apply feature scaling using StandardScaler.

```python
X = df.drop('target', axis=1)
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

scaler= StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) 
X_test_scaled = scaler.transform(X_test)      

print("Shape of scaled training data:", X_train_scaled.shape)
print("Shape of scaled testing data:", X_test_scaled.shape)
```

##### Step 7: Train Baseline Models 
##### Train three baseline classifiers: DecisionTree, RandomForest, and SVM.

```python
# Decision Tree
clf_tree = DecisionTreeClassifier(random_state=43)                      # Initialize DecisionTreeClassifier with random_state=42

# Fit the model
clf_tree.fit(X_train_scaled,y_train)                      # Fit the model on the training data (X_train, y_train)

# Predict the model
y_predict_tree= clf_tree.predict(X_test_scaled)    # Use the trained model to predict on X_test
```

```python
# Print Decision Tree results
print("Decision Tree Results:")

# Print accuracy score (use y_test and the predictions from Decision Tree)
print("Accuracy:", accuracy_score(y_test,y_predict_tree))

# Print detailed classification report (use y_test and the predictions from Decision Tree)
print(classification_report(y_test,y_predict_tree))
```

```python
# Random Forest
clf_rf = RandomForestClassifier(random_state=43)                                      # Initialize RandomForestClassifier with random_state=42


# Fit the model
clf_rf.fit(X_train_scaled,y_train)   # Fit the model on the training data (X_train, y_train)


# Predict the model
y_predict_tree=clf_rf.predict(X_test_scaled)                                     # Predict on X_test
```

```python
# Print Random Forest results
print("Random Forest Results:")

print("Accuracy:", accuracy_score(y_test,y_predict_tree)) # Use y_test and predictions

print(classification_report(y_test,y_predict_tree))
```

```python
# SVM
clf_svm = SVC(random_state=43)                             # Initialize SVC


# Fit the model
clf_svm.fit(X_train_scaled,y_train)                                        # Fit the model on the training data (X_train, y_train)


# Predict the model
y_pred_svm = clf_svm.predict(X_test_scaled)                                     # Predict on X_test
```

```python
print("SVM Results:")
print("Accuracy:", accuracy_score(y_test, y_pred_svm))                       
print(classification_report(y_test, y_pred_svm))                              
```

##### Step 8: Hyperparameter Tuning 
##### Use GridSearchCV to tune hyperparameters of one model (SVM).


```python
param_grid = {
'C': [0.1, 1, 10],
'kernel': ['linear', 'rbf']
}

svc = SVC()
grid = GridSearchCV(svc, param_grid, cv=5)


# Fit the grid search
grid.fit(X_train_scaled, y_train)                                        # Fit on training data (X_train, y_train)
```

```python
# Print best parameters and best score
print("Best Params:", grid.best_params_ )                            # grid.best_params_
print("Best CV Score:", grid.best_score_)                           # grid.best_score_
```

##### Step 9: Final Evaluation 
##### Evaluate the best model from GridSearch on the test set.



```python
# Print final tuned model results
print("Final Tuned SVM Results:")
print("Accuracy:", accuracy_score(y_test, y_pred_svm))                          # Use y_test and predictions
print(classification_report(y_test, y_pred_svm))                                # Use y_test and predictions
```

```python
import joblib

# Save the best model
joblib.dump(grid.best_estimator_, r"streamlit_and_deployment/best_model.pkl")

# Save the scaler
joblib.dump(scaler, r"streamlit_and_deployment/scaler.pkl")

print("✅ Model and scaler saved successfully!")
```
